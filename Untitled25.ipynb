```python
import os
import pandas as pd
from datetime import datetime, timedelta
from alpaca.data.historical import CryptoHistoricalDataClient
from alpaca.data.requests import CryptoBarsRequest
from alpaca.data.timeframe import TimeFrame, TimeFrameUnit

# ---------------- 1. CONFIGURATION ----------------
SYMBOL = "BTC/USD"
TIMEFRAME = TimeFrame(15, TimeFrameUnit.Minute)  # Corrected to use TimeFrame constructor for 15-min bars
DAYS_TO_FETCH = 100  # Adjustable for testing


if not API_KEY_ID or not SECRET_KEY:
    raise ValueError("Set environment variables ALPACA_API_KEY and ALPACA_SECRET_KEY")

end_date = datetime.utcnow()
start_date = end_date - timedelta(days=DAYS_TO_FETCH)

# ---------------- 2. INITIALIZE DATA CLIENT ----------------
data_client = CryptoHistoricalDataClient(
    api_key=API_KEY_ID,
    secret_key=SECRET_KEY
)

# ---------------- 3. CREATE BARS REQUEST ----------------
request_params = CryptoBarsRequest(
    symbol_or_symbols=[SYMBOL],
    timeframe=TIMEFRAME,
    start=start_date.isoformat(),
    end=end_date.isoformat()
)

# ---------------- 4. FETCH DATA ----------------
bars = data_client.get_crypto_bars(request_params)

# Convert multi-index DataFrame to single-level index
df_bars = bars.df
df_bars = df_bars.droplevel(level=0)  # drop symbol level

# Normalize timestamps to UTC and round to nearest 15-min
df_bars.index = pd.to_datetime(df_bars.index, utc=True).floor('15T')

# ---------------- 5. CHECK FOR MISSING CANDLES ----------------
expected_index = pd.date_range(
    start=df_bars.index.min(),
    end=df_bars.index.max(),
    freq='15T',
    tz='UTC'
)

missing_candles = expected_index.difference(df_bars.index)

if len(missing_candles) > 0:
    print(f"⚠️ Missing {len(missing_candles)} 15-min candles!")
    print(missing_candles)
else:
    print("✅ No missing 15-min candles detected")

# ---------------- 6. ALIGN WITH 1H DATA ----------------
# Optional: drop 15-min candles outside 1H coverage
hourly_df = pd.read_csv("btc_usd_hourly.csv", parse_dates=['timestamp'], index_col='timestamp')
hourly_index = hourly_df.index

# Corrected filtering: create a boolean mask for the index
df_bars = df_bars[(df_bars.index >= hourly_index.min()) & (df_bars.index <= hourly_index.max())]

# ---------------- 7. SUMMARY ----------------
print("--- BTC/USD 15-Minute Data Summary ---")
print(f"Timeframe: {TIMEFRAME.value}")
print(f"Data Points: {len(df_bars)}")
print(f"Date Range: {df_bars.index.min()} → {df_bars.index.max()}")
print("\nFirst 5 Rows (OHLCV):")
print(df_bars.head())
print("------------------------------------ fantasies")

# ---------------- 8. SAVE FOR BACKTEST ----------------
df_bars.to_csv("btc_usd_15min.csv")
```

```python
"""
Title: BTC/USD 1H Regime Detection
Content-Type: Python
UUID: 3d2f4e10-6c4b-4e5a-8d62-2b1f3e7a9e4f
Description: Adds Trend / Expansion / Chaos labels to 1H BTC/USD data using EMA slope and ATR.
"""

import pandas as pd
import numpy as np

# Load 1H dataset
df = pd.read_csv("btc_usd_hourly.csv", parse_dates=['timestamp'], index_col='timestamp')

# ---------------- 1. CALCULATE INDICATORS ----------------
EMA_FAST = 50
EMA_SLOW = 200
ATR_PERIOD = 14

df['EMA50'] = df['close'].ewm(span=EMA_FAST, adjust=False).mean()
df['EMA200'] = df['close'].ewm(span=EMA_SLOW, adjust=False).mean()

# True Range
df['H-L'] = df['high'] - df['low']
df['H-PC'] = abs(df['high'] - df['close'].shift(1))
df['L-PC'] = abs(df['low'] - df['close'].shift(1))
df['TR'] = df[['H-L', 'H-PC', 'L-PC']].max(axis=1)

# ATR
df['ATR'] = df['TR'].rolling(ATR_PERIOD).mean()
df['ATR_pct'] = df['ATR'] / df['close'] * 100  # as % of price

# EMA50 slope (% per candle)
df['EMA50_slope'] = df['EMA50'].pct_change() * 100

# ---------------- 2. DEFINE REGIME ----------------
def regime_label(row):
    # Trend
    if abs(row['EMA50_slope']) > 0.05:
        return 'Trend_Up' if row['EMA50_slope'] > 0 else 'Trend_Down'
    # Expansion
    elif row['ATR_pct'] > 1.5:
        return 'Expansion'
    # Chaos
    elif row['ATR_pct'] < 0.5:
        return 'Chaos'
    else:
        return 'Neutral'

df['Regime'] = df.apply(regime_label, axis=1)

# ---------------- 3. SAVE ----------------
df.to_csv("btc_usd_1h_regime.csv")

# ---------------- 4. QUICK CHECK ----------------
print("--- 1H Regime Detection Summary ---")
print(df['Regime'].value_counts())
print(df[['close', 'EMA50', 'ATR_pct', 'EMA50_slope', 'Regime']].tail())
```

```python
"""
Title: BTC/USD 15-min Regime-Gated Strategy Skeleton
Content-Type: Python
UUID: e1c5a3f7-8a4b-4c27-a84e-7d92b8f2c5d1
Description: Implements a 15-min trading strategy based on 1H regimes,
momentum (Stochastic RSI), EMA trend, and ATR volatility. Outputs backtest-ready trades.
"""

import pandas as pd
import numpy as np

# ---------------- 1. LOAD DATA ----------------
df = pd.read_csv("btc_usd_15min_with_regime.csv", parse_dates=['timestamp'], index_col='timestamp')

# ---------------- 2. INDICATORS ----------------
EMA_FAST = 15
EMA_SLOW = 50
ATR_PERIOD = 14
STOCH_RSI_PERIOD = 14

# EMA
df['EMA15'] = df['close'].ewm(span=EMA_FAST, adjust=False).mean()
df['EMA50'] = df['close'].ewm(span=EMA_SLOW, adjust=False).mean()

# True Range & ATR
df['H-L'] = df['high'] - df['low']
df['H-PC'] = abs(df['high'] - df['close'].shift(1))
df['L-PC'] = abs(df['low'] - df['close'].shift(1))
df['TR'] = df[['H-L','H-PC','L-PC']].max(axis=1)
df['ATR'] = df['TR'].rolling(ATR_PERIOD).mean()

# Stochastic RSI
delta = df['close'].diff()
up = delta.clip(lower=0)
down = -1 * delta.clip(upper=0)
roll_up = up.rolling(STOCH_RSI_PERIOD).mean()
roll_down = down.rolling(STOCH_RSI_PERIOD).mean()
RS = roll_up / roll_down.replace(0,np.nan)
df['Stoch_RSI'] = RS / RS.max()

# ---------------- 3. STRATEGY SIGNALS ----------------
def signal(row):
    # Only trade in allowed regimes
    if row['1H_Regime'] not in ['Trend_Up','Trend_Down','Expansion']:
        return 0  # no trade

    # Long
    if row['Stoch_RSI'] < 0.2 and row['EMA15'] > row['EMA50']:
        return 1
    # Short
    elif row['Stoch_RSI'] > 0.8 and row['EMA15'] < row['EMA50']:
        return -1
    else:
        return 0

df['Signal'] = df.apply(signal, axis=1)

# ---------------- 4. RISK MANAGEMENT ----------------
RISK_PER_TRADE = 0.005  # 0.5% of equity
# Placeholder: store risk per trade column for backtesting
df['Risk'] = RISK_PER_TRADE

# ---------------- 5. OUTPUT BACKTEST-READY FILE ----------------
df.to_csv("btc_usd_15min_strategy.csv")

# ---------------- 6. QUICK CHECK ----------------
print("--- Strategy Skeleton Summary ---")
print(df['Signal'].value_counts())
print(df[['close','EMA15','EMA50','ATR','Stoch_RSI','1H_Regime','Signal']].tail())
```

```python
"""
Title: 15-min Strategy Backtester
Content-Type: Python
UUID: d92c3f61-6e4b-4f72-b8e9-3e7f5f42c4a1
Description: Executes trades from the Day 4 strategy skeleton and computes PnL,
equity curve, and key metrics.
"""

import pandas as pd
import numpy as np

# ---------------- 1. LOAD STRATEGY DATA ----------------
df = pd.read_csv("btc_usd_15min_strategy.csv", parse_dates=['timestamp'], index_col='timestamp')

# Backtest parameters
INITIAL_EQUITY = 10000  # starting account size
RISK_PER_TRADE = 0.005  # 0.5% per trade
R_MULTIPLE = 2          # Take-profit = 2R
STOP_MULTIPLE = 1       # Stop-loss = 1R

df['Position'] = 0  # current open position: 1 = long, -1 = short, 0 = flat
df['Entry_Price'] = np.nan
df['Exit_Price'] = np.nan
df['PnL'] = 0.0
df['Equity'] = INITIAL_EQUITY
```

```python
equity = INITIAL_EQUITY
open_trade = None

for i, row in df.iterrows():
    signal = row['Signal']
    atr = row['ATR']
    close = row['close']

    # Check for open trade
    if open_trade is None and signal != 0:
        # Open a new trade
        position = signal
        risk_amount = equity * RISK_PER_TRADE
        stop_loss = close - position * atr * STOP_MULTIPLE
        take_profit = close + position * atr * R_MULTIPLE
        size = risk_amount / (atr * STOP_MULTIPLE)  # units of BTC
        open_trade = {
            'position': position,
            'entry': close,
            'size': size,
            'stop': stop_loss,
            'target': take_profit
        }
        df.at[i, 'Position'] = position
        df.at[i, 'Entry_Price'] = close

    elif open_trade is not None:
        # Check stop-loss / take-profit
        hit_tp = (row['high'] >= open_trade['target'] if open_trade['position'] == 1 else row['low'] <= open_trade['target'])
        hit_sl = (row['low'] <= open_trade['stop'] if open_trade['position'] == 1 else row['high'] >= open_trade['stop'])

        if hit_tp or hit_sl:
            exit_price = open_trade['target'] if hit_tp else open_trade['stop']
            pnl = open_trade['position'] * (exit_price - open_trade['entry']) * open_trade['size']
            equity += pnl
            df.at[i, 'Exit_Price'] = exit_price
            df.at[i, 'PnL'] = pnl
            df.at[i, 'Equity'] = equity
            open_trade = None
        else:
            df.at[i, 'Position'] = open_trade['position']
            df.at[i, 'Entry_Price'] = open_trade['entry']
            df.at[i, 'Equity'] = equity
    else:
        df.at[i, 'Equity'] = equity
```

```python
# Filter only completed trades
trades = df.dropna(subset=['Exit_Price'])
wins = trades[trades['PnL'] > 0]
losses = trades[trades['PnL'] <= 0]

total_trades = len(trades)
win_rate = len(wins) / total_trades if total_trades > 0 else 0
profit_factor = wins['PnL'].sum() / abs(losses['PnL'].sum()) if len(losses) > 0 else np.inf
total_return = df['Equity'].iloc[-1] - INITIAL_EQUITY
max_drawdown = (df['Equity'].cummax() - df['Equity']).max()
expectancy = trades['PnL'].mean() if total_trades > 0 else 0
sharpe = trades['PnL'].mean() / trades['PnL'].std() * np.sqrt(252*24*4) if trades['PnL'].std() != 0 else 0  # 15-min periods annualized approx

print(f"Total Trades: {total_trades}")
print(f"Win Rate: {win_rate*100:.2f}%")
print(f"Profit Factor: {profit_factor:.2f}")
print(f"Total Return: ${total_return:.2f}")
print(f"Max Drawdown: ${max_drawdown:.2f}")
print(f"Expectancy: ${expectancy:.2f}")
print(f"Sharpe Ratio (annualized): {sharpe:.2f}")
```

```python
"""
Title: BTC/USD 15-min Strategy Optimization & Analysis
Content-Type: Python
UUID: 6f3b4a27-2e49-4f0c-bc2c-9d3a8f7a1234
Description: Performs parameter grid search, tracks per-regime metrics, and plots equity curves.
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from itertools import product

# Load 15-min data
df_raw = pd.read_csv("btc_usd_15min_with_regime.csv", parse_dates=['timestamp'], index_col='timestamp')

# Parameter ranges
ema_fast_range = [10, 15, 20]
ema_slow_range = [40, 50, 60]
stoch_long_range = [0.15, 0.2, 0.25]
stoch_short_range = [0.75, 0.8, 0.85]
atr_stop_mult = [0.8, 1, 1.2]
atr_tp_mult = [1.5, 2, 2.5]

results = []

# Grid search
for ema_fast, ema_slow, stoch_long, stoch_short, stop_mult, tp_mult in product(
        ema_fast_range, ema_slow_range, stoch_long_range, stoch_short_range, atr_stop_mult, atr_tp_mult):

    df = df_raw.copy()

    # EMA
    df['EMA_fast'] = df['close'].ewm(span=ema_fast, adjust=False).mean()
    df['EMA_slow'] = df['close'].ewm(span=ema_slow, adjust=False).mean()

    # ATR
    df['H-L'] = df['high'] - df['low']
    df['H-PC'] = abs(df['high'] - df['close'].shift(1))
    df['L-PC'] = abs(df['low'] - df['close'].shift(1))
    df['TR'] = df[['H-L','H-PC','L-PC']].max(axis=1)
    df['ATR'] = df['TR'].rolling(14).mean()

    # Stochastic RSI
    delta = df['close'].diff()
    up = delta.clip(lower=0)
    down = -1 * delta.clip(upper=0)
    roll_up = up.rolling(14).mean()
    roll_down = down.rolling(14).mean()
    RS = roll_up / roll_down.replace(0,np.nan)
    df['Stoch_RSI'] = RS / RS.max()

    # Signals
    def signal(row):
        if row['1H_Regime'] not in ['Trend_Up','Trend_Down','Expansion']:
            return 0
        if row['Stoch_RSI'] < stoch_long and row['EMA_fast'] > row['EMA_slow']:
            return 1
        elif row['Stoch_RSI'] > stoch_short and row['EMA_fast'] < row['EMA_slow']:
            return -1
        else:
            return 0

    df['Signal'] = df.apply(signal, axis=1)

    # Placeholder backtest (simplified PnL tracking)
    equity = 10000
    df['PnL'] = 0.0
    df['Equity'] = float(INITIAL_EQUITY)
    open_trade = None

    for i, row in df.iterrows():
        signal_val = row['Signal']
        atr = row['ATR']
        close = row['close']
        if open_trade is None and signal_val != 0:
            pos = signal_val
            size = (equity*0.005)/(atr*stop_mult)
            stop = close - pos*atr*stop_mult
            tp = close + pos*atr*tp_mult
            open_trade = {'pos': pos, 'entry': close, 'size': size, 'stop': stop, 'tp': tp}
        elif open_trade is not None:
            hit_tp = (row['high'] >= open_trade['tp'] if open_trade['pos']==1 else row['low'] <= open_trade['tp'])
            hit_sl = (row['low'] <= open_trade['stop'] if open_trade['pos']==1 else row['high'] >= open_trade['stop'])
            if hit_tp or hit_sl:
                exit_price = open_trade['tp'] if hit_tp else open_trade['stop']
                pnl = open_trade['pos']*(exit_price - open_trade['entry'])*open_trade['size']
                equity += pnl
                df.at[i,'PnL'] = pnl
                open_trade = None
            df.at[i,'Equity'] = equity
        else:
            df.at[i,'Equity'] = equity

    # Compute summary metrics
    trades = df.dropna(subset=['PnL'])
    wins = trades[trades['PnL']>0]
    losses = trades[trades['PnL']<=0]
    total_trades = len(trades)
    win_rate = len(wins)/total_trades if total_trades>0 else 0
    profit_factor = wins['PnL'].sum()/abs(losses['PnL'].sum()) if len(losses)>0 else np.inf
    total_return = df['Equity'].iloc[-1]-10000
    max_dd = (df['Equity'].cummax()-df['Equity']).max()
    results.append({
        'ema_fast': ema_fast,
        'ema_slow': ema_slow,
        'stoch_long': stoch_long,
        'stoch_short': stoch_short,
        'stop_mult': stop_mult,
        'tp_mult': tp_mult,
        'total_trades': total_trades,
        'win_rate': win_rate,
        'profit_factor': profit_factor,
        'total_return': total_return,
        'max_drawdown': max_dd
    })

# Convert results to DataFrame
results_df = pd.DataFrame(results)
results_df.to_csv("optimization_results.csv", index=False)

# Optional: Top 5 parameter sets by total return
print(results_df.sort_values('total_return', ascending=False).head())
```

```python
"""
Title: BTC/USD 15-min Strategy Backtester (Correct Trade Accounting)
Content-Type: Python
UUID: 9c21c7b4-2b84-4f61-bc9c-6d7e91c8a4e2
Description: Proper trade-based backtester with regime gating and correct metrics.
"""

import pandas as pd
import numpy as np

# =========================
# CONFIG
# =========================
INITIAL_EQUITY = 10_000
RISK_PER_TRADE = 0.005  # 0.5%

EMA_FAST = 10
EMA_SLOW = 60
STOCH_LONG = 0.25
STOCH_SHORT = 0.80
ATR_STOP = 0.8
ATR_TP = 1.5

ALLOWED_REGIMES = ["Trend_Up", "Trend_Down", "Expansion"]

# =========================
# LOAD DATA
# =========================
df = pd.read_csv(
    "btc_usd_15min_with_regime.csv",
    parse_dates=["timestamp"],
    index_col="timestamp"
)

# =========================
# INDICATORS
# =========================
df["EMA_fast"] = df["close"].ewm(span=EMA_FAST, adjust=False).mean()
df["EMA_slow"] = df["close"].ewm(span=EMA_SLOW, adjust=False).mean()

# ATR
hl = df["high"] - df["low"]
hpc = (df["high"] - df["close"].shift()).abs()
lpc = (df["low"] - df["close"].shift()).abs()
df["TR"] = pd.concat([hl, hpc, lpc], axis=1).max(axis=1)
df["ATR"] = df["TR"].rolling(14).mean()

# Stochastic RSI (simplified, stable)
delta = df["close"].diff()
gain = delta.clip(lower=0).rolling(14).mean()
loss = (-delta.clip(upper=0)).rolling(14).mean()
rs = gain / loss.replace(0, np.nan)
df["Stoch_RSI"] = rs / rs.rolling(14).max()

df.dropna(inplace=True)

# =========================
# BACKTEST ENGINE
# =========================
equity = float(INITIAL_EQUITY)
equity_curve = []
trade_log = []

open_trade = None

for ts, row in df.iterrows():
    equity_curve.append(equity)

    # -----------------
    # ENTRY LOGIC
    # -----------------
    if open_trade is None:
        if row["1H_Regime"] in ALLOWED_REGIMES:

            if row["Stoch_RSI"] < STOCH_LONG and row["EMA_fast"] > row["EMA_slow"]:
                direction = 1
            elif row["Stoch_RSI"] > STOCH_SHORT and row["EMA_fast"] < row["EMA_slow"]:
                direction = -1
            else:
                continue

            risk_amount = equity * RISK_PER_TRADE
            stop_distance = row["ATR"] * ATR_STOP
            position_size = risk_amount / stop_distance

            open_trade = {
                "entry_time": ts,
                "entry_price": row["close"],
                "direction": direction,
                "size": position_size,
                "stop": row["close"] - direction * stop_distance,
                "tp": row["close"] + direction * row["ATR"] * ATR_TP,
                "regime": row["1H_Regime"],
            }
        continue

    # -----------------
    # EXIT LOGIC
    # -----------------
    hit_tp = (
        row["high"] >= open_trade["tp"]
        if open_trade["direction"] == 1
        else row["low"] <= open_trade["tp"]
    )

    hit_sl = (
        row["low"] <= open_trade["stop"]
        if open_trade["direction"] == 1
        else row["high"] >= open_trade["stop"]
    )

    if hit_tp or hit_sl:
        exit_price = open_trade["tp"] if hit_tp else open_trade["stop"]
        pnl = (
            open_trade["direction"]
            * (exit_price - open_trade["entry_price"])
            * open_trade["size"]
        )

        equity += pnl

        trade_log.append({
            "entry_time": open_trade["entry_time"],
            "exit_time": ts,
            "direction": open_trade["direction"],
            "regime": open_trade["regime"],
            "pnl": pnl,
            "equity_after": equity,
        })

        open_trade = None

# =========================
# METRICS
# =========================
trades = pd.DataFrame(trade_log)

total_trades = len(trades)
wins = trades[trades["pnl"] > 0]
losses = trades[trades["pnl"] < 0]

win_rate = len(wins) / total_trades if total_trades else 0
profit_factor = wins["pnl"].sum() / abs(losses["pnl"].sum()) if len(losses) else np.inf
expectancy = trades["pnl"].mean() if total_trades else 0

equity_series = pd.Series(equity_curve, index=df.index[:len(equity_curve)])
max_drawdown = (equity_series.cummax() - equity_series).max()

# =========================
# OUTPUT
# =========================
print("===== BACKTEST RESULTS =====")
print(f"Total Trades: {total_trades}")
print(f"Win Rate: {win_rate:.2%}")
print(f"Profit Factor: {profit_factor:.2f}")
print(f"Expectancy per Trade: ${expectancy:.2f}")
print(f"Total Return: ${equity - INITIAL_EQUITY:.2f}")
print(f"Max Drawdown: ${max_drawdown:.2f}")

print("\nTrades by Regime:")
print(trades.groupby("regime")["pnl"].agg(["count", "sum", "mean"]))
```
